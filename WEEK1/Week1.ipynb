## Tic-Tac-Toe Game

```python
def create_board():
    return [[' ' for _ in range(3)] for _ in range(3)]

board = create_board()
display(board)
```

```python
def display_board(board):
    for row in board:
        print('|' + '|'.join(row) + '|')
        print('-' * 7)

# Example:
# display_board(board)
```

```python
def get_player_move(player, board):
    while True:
        try:
            row = int(input(f"Player {player}, enter row (0-2): "))
            col = int(input(f"Player {player}, enter column (0-2): "))
            if 0 <= row <= 2 and 0 <= col <= 2 and board[row][col] == ' ':
                return row, col
            else:
                print("Invalid move. Please choose an empty cell within the board boundaries.")
        except ValueError:
            print("Invalid input. Please enter numbers for row and column.")
```

```python
def check_win(board, player):
    # Check rows
    for row in board:
        if all(cell == player for cell in row):
            return True
    # Check columns
    for col in range(3):
        if all(board[row][col] == player for row in range(3)):
            return True
    # Check diagonals
    if all(board[i][i] == player for i in range(3)) or \
       all(board[i][2 - i] == player for i in range(3)):
        return True
    return False
```

```python
def check_draw(board):
    return all(cell != ' ' for row in board for cell in row)
```

```python
def play_game():
    board = create_board()
    current_player = 'X'
    game_over = False

    print("Welcome to Tic-Tac-Toe!")
    player_name = ""
    player_usn = ""

    while not game_over:
        display_board(board)
        row, col = get_player_move(current_player, board)
        board[row][col] = current_player

        if check_win(board, current_player):
            display_board(board)
            print(f"Player {current_player} wins!")
            game_over = True
        elif check_draw(board):
            display_board(board)
            print("It's a draw!")
            game_over = True
        else:
            current_player = 'O' if current_player == 'X' else 'X'

    print("Game Over!"0)
    player_name = input("Please enter your name: ")
    player_usn = input("Please enter your USN: ")
    print(f"Thank you for playing, {player_name} ({player_usn}).")

# To play the game, uncomment the line below and run the cell:
play_game()
```

## Vacuum Cleaner Problem

```python
import random

SIZE = 4
vacuum = [0, 0]
dirt_piles = set()
while len(dirt_piles) < 4:
    pos = (random.randint(0, SIZE-1), random.randint(0, SIZE-1))
    if pos != tuple(vacuum):
        dirt_piles.add(pos)

score = 0
print("--- VACUUM CLEANER GRID GAME ---")
print("Controls: W (Up), S (Down), A (Left), D (Right), C (Clean), Q (Quit)\n")

while True:
    for r in range(SIZE):
        row = []
        for c in range(SIZE):
            cell = ""
            if [r, c] == vacuum:
                cell += "🤖"
            if (r, c) in dirt_piles:
                cell += "✨"
            row.append(cell if cell else "·")
        print("  ".join(row).center(20))

    print(f"\nScore (Dirt Cleaned): {score} | Remaining Dirt: {len(dirt_piles)}")
    if not dirt_piles:
        print("🎉 Congratulations! You cleaned all the dirt!")
        break

    action = input("Enter move (W/A/S/D) or Clean (C): ").strip().upper()

    if action == 'Q':
        print("Game Over. Thanks for playing!")
        break
    elif action == 'C':
        if tuple(vacuum) in dirt_piles:
            dirt_piles.remove(tuple(vacuum))
            score += 1
            print("🧹 Cleaned the room!")
        else:
            print("❌ No dirt here to clean.")
    elif action in ['W', 'A', 'S', 'D']:
        if action == 'W' and vacuum[0] > 0: vacuum[0] -= 1
        elif action == 'S' and vacuum[0] < SIZE - 1: vacuum[0] += 1
        elif action == 'A' and vacuum[1] > 0: vacuum[1] -= 1
        elif action == 'D' and vacuum[1] < SIZE - 1: vacuum[1] += 1
        else:
            print("💥 Bumped into a wall!")
    else:
        print("⚠️ Invalid key! Use W, A, S, D, C, or Q.")
    print("-" * 30)

player_name = input("Please enter your name: ")
player_usn = input("Please enter your USN: ")
print(f"Thank you for playing, {player_name} ({player_usn}).")
```
